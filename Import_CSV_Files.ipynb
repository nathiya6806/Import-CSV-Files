```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
columns = [
    "SepalLength",
    "SepalWidth",
    "PetalLength",
    "PetalWidth",
    "Species"
]

df = pd.read_csv(
    "/content/iris.data",
    header=None,
    names=columns
)

df = df.dropna()

df.head()
```

```python
print("Dataset Information:")
df.info()

print("\nDataset Shape:")
print(df.shape)

print("\nStatistical Summary:")
display(df.describe())
```

```python
print("First 5 Rows:")
display(df.head())

print("Last 5 Rows:")
display(df.tail())

print("Missing Values:")
print(df.isnull().sum())

print("Species Count:")
print(df["Species"].value_counts())
```

```python
df["Species"].value_counts().plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Iris Species Distribution")
plt.xlabel("Species")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["PetalLength"],
    df["PetalWidth"]
)

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")

plt.show()
```
